```python
def insertion_sort(A):
    n = len(A)
    for j in range(1, n):
        key = A[j]
        i = j - 1
        while i >= 0 and A[i][1] > key[1]:
            A[i + 1] = A[i]
            i = i - 1
        A[i + 1] = key
    return A

rank_list = [
    ("Alice", 85),
    ("Bob", 92),
    ("Charlie", 88),
    ("David", 91),
    ("Eva", 78),
]
sorted_rank_list = insertion_sort(rank_list)[::-1]
print("Sorted Admission Test Rank List:")
for rank, (student, score) in enumerate(sorted_rank_list, start=1):
    print(f"Rank {rank}: {student} - {score}")
```

```python
def insertion_sort(A):
    n = len(A)
    # Start sorting from the second element (index 1)
    for j in range(1, n):
        key = A[j]  # Key is the current student tuple
        i = j - 1
        # Move elements of A[0..i] that are smaller than key[1] (the score) to one position ahead of their current position
        while i >= 0 and A[i][1] > key[1]:  # Compare based on the second element of the tuple (score))
            A[i + 1] = A[i]
            i = i - 1
        A[i + 1] = key  # Place key at the correct position
    return A

# Example list of students and their scores
rank_list = [
    ("Alice", 85),
    ("Bob", 92),
    ("Charlie", 88),
    ("David", 91),
    ("Eva", 78),
]
sorted_rank_list = insertion_sort(rank_list)[::-1]
# Display sorted rank list in descending order
print("Sorted Admission Test Rank List:")
for rank, (student, score) in enumerate(sorted_rank_list, start=1):
    print(f"Rank {rank}: {student} - {score}")
```
